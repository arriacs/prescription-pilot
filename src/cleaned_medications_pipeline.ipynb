# Cleaning Up Date

```python
#normalizing by 
TEXT_COLS = [
    "title", "brand_names", "generic_names", "active_ingredients",
    "manufacturer", "dosage_and_administration",
    "drug_interactions", "adverse_reactions",
]
```

```python
for col in TEXT_COLS:
    df[col] = (
        df[col].astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .replace("", pd.NA)
    )
```

```python
df
```

```python
artifact = r"Error! Hyperlink reference not valid\.?"
for col in ["dosage_and_administration", "drug_interactions", "adverse_reactions"]:
    df[col] = df[col].str.replace(artifact, "", regex=True).str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df
```

```python
heading_patterns = {
    "dosage_and_administration": r"^(?:\d+\s+)?DOSAGE AND ADMINISTRATION\s*[:—-]*\s*",
    "drug_interactions": r"^(?:\d+\s+)?(?:DRUG INTERACTIONS|Drug Interactions)\s*[:—-]*\s*",
    "adverse_reactions": r"^(?:\d+\s+)?ADVERSE REACTIONS\s*[:—-]*\s*",
}

for col, pat in heading_patterns.items():
    df[col] = df[col].str.replace(pat, "", regex=True, flags=re.IGNORECASE)
```

```python
df
```

```python
#standardize IDs and dates
df["set_id"] = df["set_id"].astype("string").str.strip().str.lower()
df["effective_date"] = pd.to_datetime(
    df["effective_date"].astype("string"), format="%Y%m%d", errors="coerce"
)
```

```python
df["effective_date"]
```

```python
boiler = r"^These highlights do not include all the information needed to use .*? safely and effectively\.\s*See full prescribing information for .*?\.\s*"
df["title"] = (
    df["title"].str.replace(boiler, "", regex=True)
    .str.replace(r"[®™]", "", regex=True)
    .str.strip()
    .replace("", pd.NA)
)
```

```python
df
```

```python
#identifying the columns that had missing information
df["title"] = df["title"].fillna(df["brand_names"]).fillna(df["generic_names"])
df["active_ingredients"] = df["active_ingredients"].fillna(df["generic_names"].str.upper())
df["brand_names"] = df["brand_names"].fillna(df["generic_names"])  # fallback; see brand_is_generic
df["brand_is_generic"] = df["brand_names"].str.lower() == df["generic_names"].str.lower()
```

```python
df = (
    df.sort_values(["set_id", "version", "effective_date"])
    .drop_duplicates("set_id", keep="last")
)
```

```python
df
```

```python
#Removing duplicated rows
dedupe_cols = ["generic_names", "manufacturer", "dosage_and_administration",
               "drug_interactions", "adverse_reactions"]
before = len(df)
df = df.sort_values("effective_date").drop_duplicates(dedupe_cols, keep="last")
print(f"Dropped {before - len(df)} content-duplicate rows")
 
```

```python
df["missing_key_text"] = df[["dosage_and_administration", "adverse_reactions"]].isna().any(axis=1)
 
df = df.reset_index(drop=True)
 
print("Cleaned:", df.shape)
print("\nMissing values after cleaning:")
print(df.isna().sum())
 
df.to_csv(OUT_PATH, index=False)
print("\nSaved to", OUT_PATH)
```
